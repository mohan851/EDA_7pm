```python
plt.figure(figsize=(25,15))
for i in range(1,len(cat)):
    keys = visadf[cat[i]].value_counts().keys().to_list()
    values = visadf[cat[i]].value_counts().values.tolist()
    plt.subplot(3,3,i)
    plt.bar(x = keys , height=values)
    plt.xlabel("continents")
    plt.ylabel("population")
    plt.title(cat[i])


```

```python
plt.figure(figsize=(25,15))
for i,c in enumerate(cat[1:]):
    
    keys = visadf[c].value_counts().keys().to_list()
    values = visadf[c].value_counts().values.tolist()
    plt.subplot(3,3,i+1)
    plt.bar(x = keys , height=values)
    plt.xlabel("continents")
    plt.ylabel("population")
    plt.title(c)
```
