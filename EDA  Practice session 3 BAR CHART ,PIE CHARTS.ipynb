# Bar Chart

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
path = r"C:\Users\Admin\Documents\naresh IT\data packages\Visadataset.csv"
visadf = pd.read_csv(path)
```

```python
visadf
```

```python
keys = visadf['continent'].value_counts().keys().to_list()
```

```python
keys
```

```python
values= visadf['continent'].value_counts().values.tolist()
```

```python
values
```

```python
plt.figure(figsize=(10,6))
plt.bar(x = keys , height=values)
plt.xlabel("continents")
plt.ylabel("population")
plt.title("Contry population")
plt.savefig("Contry population.png")
plt.show()

```
